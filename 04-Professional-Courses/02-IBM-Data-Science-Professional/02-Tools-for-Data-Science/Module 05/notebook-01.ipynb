```python
# تفعيل الإضافة (دي بتتعمل مرة واحدة بس في الملف)
%load_ext rpy2.ipython

import pandas as pd
df = pd.DataFrame({'a': [1, 2, 3], 'b': [4, 5, 6]})
```

```python
%%R -i df
# دلوقتى السطر الأول فيه السحر.. هيشتغل تمام
library(ggplot2)
ggplot(df, aes(x=a, y=b)) + geom_line()
```
